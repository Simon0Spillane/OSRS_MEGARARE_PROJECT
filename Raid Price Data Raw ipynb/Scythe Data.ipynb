```python
import requests
import pandas as pd
import pandas as pd
import numpy as np
import seaborn as sb
import matplotlib.pyplot as plt
```

```python
URL = "https://api.weirdgloop.org/exchange/history/osrs/all"
PARAMS = {"id": 22486}
HEADERS = {"User-Agent": "WEBDATASCIENCE OSRS Research Project/1.0, simon.spillane@colorado.edu"}
response = requests.get(URL, params=PARAMS, headers=HEADERS)
response.raise_for_status()
print(response)

ScytheData = response.json()
records = ScytheData["22486"]
dfScythe = pd.DataFrame(
    records,
    columns=["timestamp", "price", "volume"][:len(records[0])]
)
dfScythe["date"] = pd.to_datetime(
    dfScythe["timestamp"],
    unit="ms"
)
start_date = "2018-09-26"
end_date = "2026-10-30"

filtered_df = dfScythe[
    (dfScythe["date"] >= start_date) &
    (dfScythe["date"] < "2026-10-30")
]
dfScythe.plot(
    x="date",
    y="price",
    figsize=(14, 6),
    title="Scythe of Vitur Historical GE Price",
    legend=False
);
```
