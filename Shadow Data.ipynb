```python
import requests
import pandas as pd
import pandas as pd
import numpy as np
import seaborn as sb
import matplotlib.pyplot as plt
```

```python
URL = "https://api.weirdgloop.org/exchange/history/osrs/all"
PARAMS = {"id": 27277}
HEADERS = {"User-Agent": "WEBDATASCIENCE OSRS Research Project/1.0, simon.spillane@colorado.edu"}
response = requests.get(URL, params=PARAMS, headers=HEADERS)
response.raise_for_status()
print(response)

ShadowData = response.json()
records = ShadowData["27277"]
dfShadow = pd.DataFrame(
    records,
    columns=["timestamp", "price", "volume"][:len(records[0])]
)
dfShadow["date"] = pd.to_datetime(
    dfShadow["timestamp"],
    unit="ms"
)
start_date = "2018-09-26"
end_date = "2026-10-30"

filtered_df = dfShadow[
    (dfShadow["date"] >= start_date) &
    (dfShadow["date"] < "2026-10-30")
]
dfShadow.plot(
    x="date",
    y="price",
    figsize=(14, 6),
    title="Shadow of Vitur Historical GE Price",
    legend=False
);
```

```python
from matplotlib.ticker import MultipleLocator, FuncFormatter
from matplotlib.ticker import MultipleLocator, FuncFormatter


ax = filtered_df.plot(
    x="date",
    y="price",
    figsize=(14, 6),
    title="Scythe of Vitur Historical GE Price",
    legend=False
)
ax.yaxis.set_major_locator(MultipleLocator(50_000_000))
ax.axvline(pd.to_datetime("2022-08-24"), color="b", zorder=-1)
ax.axvline(pd.to_datetime("2026-10-20"), color="g", zorder=-1)
```
